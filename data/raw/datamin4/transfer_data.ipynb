```python
import os
import pandas as pd
data_list = []
for data in os.listdir(".\data_barra"):
    data_df = pd.read_csv(f".\data_barra\{data}")
    data_df["date"] = data[:-4]
    data_list.append(data_df)
```

```python
barra_data = pd.concat(data_list,axis=0,ignore_index=True)
```

```python
barra_data
```

```python
size_wide = barra_data.pivot(index="date",columns="code",values="size")
```

```python
import numpy as np
def ts_max(x: np.ndarray, window_size: int) -> np.ndarray:
    if window_size > x.shape[0]:
        raise ValueError("The window size cannot be greater than the number of time points")
    windows = np.lib.stride_tricks.sliding_window_view(x, window_shape=(window_size,), axis=0)
    max_value = np.nanmax(windows, axis=2)
    return max_value

ts_max(size_wide.values,4)
```
